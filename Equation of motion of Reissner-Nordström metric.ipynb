# Reissner–Nordström metric

The goal of this notebook is to derive the equations of motion of the Reissner–Nordström metric. The Reissner–Nordström metric is given by (see Wikipedia: https://en.wikipedia.org/wiki/Reissner%E2%80%93Nordstr%C3%B6m_metric):

$${\displaystyle \mathrm {d} s^{2}=c^{2}\left(1-{\frac {r_{\mathrm {S} }}{r}}+{\frac {r_{\mathrm {Q} }^{2}}{r^{2}}}\right)\mathrm {d} t^{2}-\left(1-{\frac {r_{\mathrm {S} }}{r}}+{\frac {r_{\mathrm {Q} }^{2}}{r^{2}}}\right)^{-1}\mathrm {d} r^{2}-r^{2} \mathrm {d} \theta^2 - r^{2} \sin ^{2} \theta} \mathrm {d} \varphi ^2$$

With :
- ${\displaystyle r_{\text{s}}}$ is the Schwarzschild radius of the body given by ${\displaystyle \textstyle r_{\text{s}}={\frac {2GM}{c^{2}}}}$
- ${\displaystyle r_{Q}}$ is a characteristic length scale given by ${\displaystyle \textstyle r_{Q}^{2}={\frac {Q^{2}G}{4\pi \varepsilon _{0}c^{4}}}}$

It is a generalisation of the Schwarschild Metric.

For that we will use the Geodesic Equation :

$${\displaystyle {d^{2}x^{\mu } \over ds^{2}}+\Gamma ^{\mu }{}_{\alpha \beta }{dx^{\alpha } \over ds}{dx^{\beta } \over ds}=0\ }$$

### Reissner–Nordström metric

```python
Q, epsilon, s = smp.symbols("Q varepsilon_0 s")
r_s, r_Q = 2*G*M/c**2, smp.sqrt(Q**2*G / (4 * smp.pi * epsilon * c**4))
```

```python
R_N = smp.Matrix(np.zeros([4,4], dtype=int))

R_N[0,0] = 1 - r_s/r + (r_Q/r)**2
R_N[1,1] = -1 / (1 - r_s/r + (r_Q/r)**2)
R_N[2,2] = -r**2
R_N[3,3] = -r**2 * smp.sin(theta) ** 2

R_N
```

### Christoffel Symbols

```python
Gamma = Christoffel_Symbols(R_N)
Display_Christoffel_Symbols(R_N, Gamma)
```
